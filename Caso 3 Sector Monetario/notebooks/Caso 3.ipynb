```python
# Data Question 1

variable_ques1 = ["numerario en circulacion", "credito bancario total al sector privado", 
                  "medios de pagos totales", "remesas familiares", "PIB Trimestral"]
df_ques1 = df[["Año", "Trimestre"] + variable_ques1]
df_ques1["Periodo"] = (df_ques1["Año"].astype(str) + "-T"+ df_ques1["Trimestre"].astype(str))
df_ques1["Fecha"] = pd.to_datetime(df_ques1["Año"].astype(str) + "-" + ((df_ques1["Trimestre"] - 1) * 3 + 1).astype(str) + "-01")

for col in variable_ques1: 
    df[col] = (df[col].astype(str).str.replace(",", "", regex=False).astype(float))

for var in variable_ques1: 
    df_ques1[var] = df[var] / df[var].iloc[0] * 100

source = ColumnDataSource(df_ques1)
df_ques1.columns.name = None
TABLES["table_ques1"] = df_ques1
df_ques1
```

```python
#Chart Question 1
p_ques1 = figure(
    width=950,
    height=450,
    x_axis_type="datetime",
    title="Evolución de los principales agregados monetarios (Base 100)",
    background_fill_color=theme["BACKGROUND"],
    border_fill_color=theme["PANEL"],
    tools="pan,wheel_zoom,box_zoom,reset,save"
)
p_ques1 .line(
    "Fecha",
    "numerario en circulacion",
    source=source,
    color=colors["NUMERARIO"],
    line_width=3,
    legend_label="Numerario"
)

p_ques1 .line(
    "Fecha",
    "credito bancario total al sector privado",
    source=source,
    color=colors["CREDITO"],
    line_width=3,
    legend_label="Crédito"
)

p_ques1 .line(
    "Fecha",
    "medios de pagos totales",
    source=source,
    color=colors["M2"],
    line_width=3,
    legend_label="M2"
)

p_ques1 .line(
    "Fecha",
    "remesas familiares",
    source=source,
    color=colors["REMESAS"],
    line_width=3,
    legend_label="Remesas"
)

p_ques1 .line(
    "Fecha",
    "PIB Trimestral",
    source=source,
    color=colors["PIB"],
    line_width=3,
    legend_label="PIB"
)

hover = HoverTool(
    tooltips="""
    <div style="
        background:#1f1720;
        border:1px solid #ff6b81;
        border-radius:8px;
        padding:8px;
        color:white;
    ">
        <b>Fecha:</b> @Periodo<br>
        <b>Valor:</b> $y{0.0}
    </div>
    """
)

p_ques1 .add_tools(hover)
p_ques1 .stylesheets = [
    GlobalInlineStyleSheet(css="""
    .bk-Tooltip {
        --tooltip-color: transparent !important;
        --tooltip-border: transparent !important;
    }
    .bk-tooltip-content {
        background: transparent !important;
        border: none !important;
        box-shadow: none !important;
        padding: 0 !important;
    }
    """)
]
p_ques1.legend.location = "top_left"
p_ques1.legend.orientation = "vertical"
p_ques1.legend.background_fill_color = theme["PANEL"]
p_ques1.legend.background_fill_alpha = 0.9
p_ques1.legend.border_line_color = theme["BORDER"]
p_ques1.legend.label_text_color = theme["TEXT"]
p_ques1.legend.label_text_font_size = "10pt"
p_ques1.legend.padding = 8
p_ques1.legend.spacing = 6
p_ques1.legend.margin = 5
p_ques1.legend.click_policy = "hide"
p_ques1.yaxis.formatter = NumeralTickFormatter(format="0.00a")
p_ques1.toolbar.logo = None
p_ques1.toolbar.autohide = True
p_ques1.title.text_color = theme["ACCENT"] 
p_ques1.title.text_font_size = "20pt"        
p_ques1.title.align = "center"  
p_ques1.yaxis.axis_label = 'Saldos'
p_ques1.yaxis.axis_label_text_color = theme["TEXT"]
p_ques1.xaxis.major_label_text_color = theme["TEXT_SOFT"]
p_ques1.yaxis.major_label_text_color = theme["TEXT_SOFT"]
p_ques1.outline_line_color = theme["BORDER"]
p_ques1.xgrid.grid_line_color = theme["GRID"]
p_ques1.ygrid.grid_line_color = theme["GRID"]
p_ques1.xgrid.grid_line_alpha = theme["GRID_ALPHA"]
p_ques1.ygrid.grid_line_alpha = theme["GRID_ALPHA"]

select = figure(
    title="Seleccionar período",
    height=120,
    width=950,
    x_axis_type="datetime",
    y_axis_type=None,
    tools="",
    toolbar_location=None,
    background_fill_color=theme["BACKGROUND"],
    border_fill_color=theme["PANEL"]
)
select.line("Fecha", "medios de pagos totales", source=source, color=colors["M2"])
range_tool = RangeTool(x_range=p_ques1.x_range)
range_tool.overlay.fill_color = theme["ACCENT"]
range_tool.overlay.fill_alpha = 0.25
select.add_tools(range_tool)

CHARTS["chart_ques1"] = column(p_ques1)
show(column(p_ques1, select))
```

Las diferencias en el comportamiento de estas variables pueden asociarse a su naturaleza económica. Esta dependencia explica la mayor volatilidad observada en dicha variable.

No obstante, el análisis de estos indicadores debe complementarse con variables como el crédito bancario, los medios de pago y el PIB, ya que en conjunto proporcionan una visión más completa de la evolución del sistema monetario y financiero del país y permiten fundamentar de mejor manera la toma de decisiones.
