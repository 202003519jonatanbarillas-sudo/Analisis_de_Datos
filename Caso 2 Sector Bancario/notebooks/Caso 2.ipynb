```python
#Datos pregunta 1
df_ques1 = (df.groupby("mora").size().reset_index(name = "clientes"))
df_ques1["porcentaje"] = (df_ques1["clientes"]/df_ques1["clientes"].sum())
df_ques1["angulo"] = (df_ques1["porcentaje"] * 2 * pi)
df_ques1["end_angle"] = df_ques1["angulo"].cumsum()
df_ques1["start_angle"] = (df_ques1["end_angle"] - df_ques1["angulo"])
colors = {"Sí": theme["NEGATIVE"],"No": theme["POSITIVE"]}
df_ques1["colors"] = df_ques1["mora"].map(colors)
df_ques1["porcentaje_str"] = (df_ques1["porcentaje"] * 100).round(2)
source = ColumnDataSource(df_ques1)

df_ques1.columns.name = None
df_ques1
```

```python
# Gráfica pregunta 1
p = figure(
    width=650,
    height=500,
    x_range=(-1, 1),
    y_range=(-1, 1),
    background_fill_color=theme["BACKGROUND"],
    border_fill_color=theme["PANEL"],
    tools="hover, save",
    tooltips="""
    <div style="
        background:#111827;
        border:1px solid #3b82f6;
        border-radius:8px;
        padding:8px;
        color:white;
    ">
        <b>Estado:</b> @mora <br>
        <b>Clientes:</b> @clientes{0,0} <br>
        <b>Porcentaje:</b> @porcentaje_str %
    </div>
    """
)

p.annular_wedge(
    x=0,
    y=0,
    inner_radius=0.35,
    outer_radius=0.60,
    start_angle="start_angle",
    end_angle="end_angle",
    color="colors",
    legend_field="mora",
    line_color=theme["BACKGROUND"],
    line_width=3,
    source=source
)

texto = Label(
    x=0,
    y=0.05,
    x_units="data",
    y_units="data",
    text="Tasa de Mora",
    text_align="center",
    text_baseline="middle",
    text_color=theme["TEXT_SOFT"],
    text_font_size="13pt"
)

valor = Label(
    x=0,
    y=-0.1,
    x_units="data",
    y_units="data",
    text=f"{df_ques1.loc[df_ques1['mora']=='Sí','porcentaje_str'].iloc[0]}%",
    text_align="center",
    text_baseline="middle",
    text_font_size="20pt",
    text_font_style="bold",
    text_color=theme["ACCENT"]
)


p.stylesheets = [
    GlobalInlineStyleSheet(css="""
    .bk-Tooltip {
        --tooltip-color: transparent !important;
        --tooltip-border: transparent !important;
    }

    .bk-tooltip-content {
        background: transparent !important;
        border: none !important;
        box-shadow: none !important;
        padding: 0 !important;
    }
    """)
]

p.title.text = "Distribución de Clientes según Mora"
p.title.text_color = theme["ACCENT"]
p.title.text_font_size = "15pt"
p.title.align = "center"
p.legend.location = "right"
p.legend.label_text_color = theme["TEXT"]
p.legend.background_fill_color = theme["PANEL"]
p.legend.border_line_color = theme["BORDER"]
p.legend.label_text_font_size = "11pt"
p.add_layout(texto)
p.add_layout(valor)
p.axis.visible = False
p.grid.visible = False
p.outline_line_color = theme["BORDER"]
p.toolbar.logo = None
p.toolbar.autohide = True

show(p)
```

```python
#Gráfica pregunta 3
p = figure(
    height=400, width=800,
    tools="pan,wheel_zoom,reset,save,tap",
    x_axis_label="Ingreso (Q)",
    y_axis_label="Saldo disponible (Q)",
    background_fill_color=theme["BACKGROUND"],
    border_fill_color=theme["PANEL"],
    title="Relación ingreso - prestamo"
)

renderer = p.scatter(
    x="ingreso", y="prestamo",
    source=source,
    size=8,
    fill_color=factor_cmap("sector", palette=["#d4af37","#29a57a","#377ad4","#f59e0b"], factors=df["sector"].unique()),
    line_color=None
)

hover = HoverTool(tooltips="""
    <div style="background:#0b0b0b; border:1px solid #d4af37; border-radius:6px; padding:6px; color:#fff;">
        <div><b>Cliente:</b> @cliente_id</div>
        <div><b>Ingreso:</b> @ingreso{0,0}</div>
        <div><b>Prestamo:</b> @prestamo{0,0}</div>
        <div><b>Sector:</b> @sector</div>
    </div>
""", renderers=[renderer])
p.add_tools(hover)

callback = CustomJS(args=dict(source=source), code="""
    const indices = source.selected.indices;
    if (indices.length > 0) {
        const seg = source.data['sector'][indices[0]];
        const new_indices = [];
        for (let i = 0; i < source.data['sector'].length; i++) {
            if (source.data['sector'][i] === seg) {
                new_indices.push(i);
            }
        }
        source.selected.indices = new_indices;
    }
""")

taptool = p.select(type=TapTool)
taptool.callback = callback
p.yaxis.formatter = NumeralTickFormatter(format="0.00a")
p.title.text_color = theme["ACCENT"] 
p.title.text_font_size = "20pt"        
p.title.align = "center"  
p.xaxis.axis_label_text_color = theme["TEXT"]
p.yaxis.axis_label_text_color = theme["TEXT"]
p.xaxis.major_label_text_color = theme["TEXT_SOFT"]
p.yaxis.major_label_text_color = theme["TEXT_SOFT"]
p.outline_line_color = theme["BORDER"]
p.xgrid.grid_line_color = theme["GRID"]
p.ygrid.grid_line_color = theme["GRID"]
p.xgrid.grid_line_alpha = theme["GRID_ALPHA"]
p.ygrid.grid_line_alpha = theme["GRID_ALPHA"]

p.toolbar.logo = None
p.toolbar.autohide = True
show(p)
```

```python
#Datos pregunta 4
df_ques4 = df[["antiguedad_laboral","mora"]]
source = ColumnDataSource(df_ques4)
df_ques4.columns.name = None
df_ques4
```

```python
#Gráfica pregunta 4
cats = ["Sin mora", "Con mora"]

vals = [
    df[df["mora"] == "No"]["antiguedad_laboral"],
    df[df["mora"] == "Sí"]["antiguedad_laboral"]
]

p_box = figure(
    x_range=cats,
    title="Distribución de la Antigüedad Laboral según Estado de Mora",
    width=750,
    height=450,
    tools="pan,wheel_zoom,box_zoom,reset,save"
)

estadisticas = []

for i, v in enumerate(vals):
    q1 = v.quantile(0.25)
    q2 = v.median()
    q3 = v.quantile(0.75)
    mean_val = v.mean()
    iqr = q3 - q1
    upper = v[v <= q3 + 1.5 * iqr].max()
    lower = v[v >= q1 - 1.5 * iqr].min()
    outliers = v[(v > upper) | (v < lower)]
    cat = cats[i]
    color = theme["POSITIVE"] if i == 0 else theme["NEGATIVE"]
    estadisticas.append({
        "estado": cat,
        "media": mean_val,
        "mediana": q2,
        "q1": q1,
        "q3": q3,
        "min": lower,
        "max": upper
    })

    p_box.vbar(
        x=[cat],
        width=0.45,
        bottom=[q1],
        top=[q3],
        fill_color=color,
        line_color="black",
        alpha=0.75
    )

    p_box.rect(
        x=[cat],
        y=[q2],
        width=0.6,
        height=0.03,
        color="black"
    )

    p_box.scatter(
        x=[cat],
        y=[mean_val],
        marker="diamond",
        size=12,
        color="#f59e0b"
    )

    p_box.segment(
        x0=[cat],
        y0=[q3],
        x1=[cat],
        y1=[upper],
        line_width=2,
        line_color="#475569"
    )

    p_box.segment(
        x0=[cat],
        y0=[q1],
        x1=[cat],
        y1=[lower],
        line_width=2,
        line_color="#475569"
    )

    p_box.rect(
        x=[cat],
        y=[upper],
        width=0.18,
        height=0.01,
        color="#475569"
    )

    p_box.rect(
        x=[cat],
        y=[lower],
        width=0.18,
        height=0.01,
        color="#475569"
    )

    if len(outliers) > 0:
        p_box.scatter(
            x=[cat] * len(outliers),
            y=outliers.tolist(),
            size=6,
            color="#f59e0b",
            alpha=0.7
        )

p_box.toolbar.logo = None

stats_source = ColumnDataSource(pd.DataFrame(estadisticas))

hover_renderer = p_box.scatter(
        x="estado",
        y="media",
        size=40,
        alpha=0,
        source=stats_source
)

hover = HoverTool(
    renderers=[hover_renderer],
    tooltips="""
    <div style="
        background:#111827;
        border:1px solid #3b82f6;
        border-radius:8px;
        padding:10px;
        color:white;
    ">
        <b>Estado:</b> @estado<br><br>

        <b>Media:</b> @media{0.0} años<br>
        <b>Mediana:</b> @mediana{0.0} años<br>
        <b>Q1:</b> @q1{0.0} años<br>
        <b>Q3:</b> @q3{0.0} años<br>
        <b>Mínimo:</b> @min{0} años<br>
        <b>Máximo:</b> @max{0} años
    </div>
    """
)

p_box.add_tools(hover)
p_box.background_fill_color = theme["BACKGROUND"]
p_box.border_fill_color = theme["PANEL"]
p_box.outline_line_color = theme["BORDER"]
p_box.title.text_color = theme["ACCENT"]
p_box.title.text_font_size = "15pt"
p_box.title.align = "center"
p_box.yaxis.axis_label = "Antigüedad Laboral (años)"
p_box.xaxis.axis_label_text_color = theme["TEXT"]
p_box.yaxis.axis_label_text_color = theme["TEXT"]
p_box.xaxis.major_label_text_color = theme["TEXT_SOFT"]
p_box.yaxis.major_label_text_color = theme["TEXT_SOFT"]
p_box.xaxis.axis_line_color = theme["BORDER"]
p_box.yaxis.axis_line_color = theme["BORDER"]
p_box.xaxis.major_tick_line_color = theme["TEXT_SOFT"]
p_box.yaxis.major_tick_line_color = theme["TEXT_SOFT"]
p_box.xaxis.minor_tick_line_color = None
p_box.yaxis.minor_tick_line_color = None
p_box.xgrid.grid_line_color = None
p_box.ygrid.grid_line_color = theme["GRID"]
p_box.ygrid.grid_line_alpha = 0.35
p_box.toolbar.logo = None
p_box.toolbar.autohide = True
p_box.y_range.start = -1
p_box.y_range.end = df["antiguedad_laboral"].max() + 2


p_box.stylesheets = [
    GlobalInlineStyleSheet(css="""
    .bk-Tooltip{
        --tooltip-color:transparent !important;
        --tooltip-border:transparent !important;
    }

    .bk-tooltip-content{
        background:transparent !important;
        border:none !important;
        box-shadow:none !important;
        padding:0 !important;
    }
    """)
]

show(p_box)
```
