```python
# Data Question 1
df_ques1 = df.groupby("fecha", as_index=False)["saldo"].sum()
df_ques1["fecha_str"] = df_ques1["fecha"].dt.strftime("%d-%m-%Y")
source = ColumnDataSource(df_ques1)
df_ques1.columns.name = None
TABLES["table_ques1"] = df_ques1
df_ques1
```

```python
# Chart Question 1

p_ques1 = figure(height=550, 
           width=950, 
           tools="xpan,xwheel_zoom,reset,save",
           x_axis_type="datetime", 
           x_axis_location="above",
           background_fill_color=theme["BACKGROUND"],
           border_fill_color=theme["PANEL"],
           x_range=(df_ques1["fecha"].min(), df_ques1["fecha"].max()),
           title="Comportamiento de los saldos durante el año 2025"
)
p_ques1.line('fecha', 'saldo', source=source, line_width=2, color=theme["ACCENT"])
hover = HoverTool(
    tooltips=f"""
    <div style="background:{theme['BACKGROUND']}; border:1px solid {theme['ACCENT']}; border-radius:8px; padding:6px 10px; color:{theme['TEXT']};">
        <div><span style="color:{theme['ACCENT']};">Fecha:</span> @fecha_str</div>
        <div><span style="color:{theme['ACCENT']};">Depósitos:</span> @saldo{{0,0}}</div>
    </div>
    """,
    mode="vline"
)

p_ques1.add_tools(hover)

p_ques1.stylesheets = [
    GlobalInlineStyleSheet(css="""
    .bk-Tooltip {
        --tooltip-color: transparent !important;
        --tooltip-border: transparent !important;
    }
    .bk-tooltip-content {
        background: transparent !important;
        border: none !important;
        box-shadow: none !important;
        padding: 0 !important;
    }
    """)
]

p_ques1.yaxis.formatter = NumeralTickFormatter(format="0.00a")
p_ques1.toolbar.logo = None
p_ques1.toolbar.autohide = True
p_ques1.title.text_color = theme["ACCENT"] 
p_ques1.title.text_font_size = "20pt"        
p_ques1.title.align = "center"  
p_ques1.yaxis.axis_label = 'Saldos'
p_ques1.yaxis.axis_label_text_color = theme["TEXT"]
p_ques1.xaxis.major_label_text_color = theme["TEXT_SOFT"]
p_ques1.yaxis.major_label_text_color = theme["TEXT_SOFT"]
p_ques1.outline_line_color = theme["BORDER"]
p_ques1.xgrid.grid_line_color = theme["GRID"]
p_ques1.ygrid.grid_line_color = theme["GRID"]
p_ques1.xgrid.grid_line_alpha = theme["GRID_ALPHA"]
p_ques1.ygrid.grid_line_alpha = theme["GRID_ALPHA"]

select = figure(title="Rango de Temporalidad",
                height=130, width=950,
                x_axis_type="datetime", 
                y_axis_type=None,
                tools="", toolbar_location=None,
                background_fill_color=theme["BACKGROUND"],
                border_fill_color=theme["PANEL"])

range_tool = RangeTool(x_range=p_ques1.x_range)
range_tool.overlay.fill_color = theme["ACCENT"]
range_tool.overlay.fill_alpha = 0.3

select.line('fecha', 'saldo', source=source, color=theme["ACCENT"])
select.ygrid.grid_line_color = None
select.add_tools(range_tool)

CHARTS["chart_ques1"] = column(p_ques1)
show(column(p_ques1, select))
```

```python
# Data Question 4

df_ques4 = df[["cliente_id","ingreso","saldo","segmento"]]
source = ColumnDataSource(df_ques4)
df_ques4.columns.name = None
TABLES["table_ques4"] = df_ques4
df_ques4
```

```python
# Chart Question 4

p_ques4 = figure(
    height=400, width=800,
    tools="pan,wheel_zoom,reset,save,tap",
    x_axis_label="Ingreso (Q)",
    y_axis_label="Saldo disponible (Q)",
    background_fill_color=theme["BACKGROUND"],
    border_fill_color=theme["PANEL"],
    title="Relación saldo - ingreso"
)

renderer = p_ques4.scatter(
    x="ingreso", y="saldo",
    source=source,
    size=8,
    fill_color=factor_cmap("segmento", palette=["#d4af37","#29a57a","#377ad4"], factors=df_ques4["segmento"].unique()),
    line_color=None
)

hover = HoverTool(tooltips="""
    <div style="background:#0b0b0b; border:1px solid #d4af37; border-radius:6px; padding:6px; color:#fff;">
        <div><b>Cliente:</b> @cliente_id</div>
        <div><b>Ingreso:</b> @ingreso{0,0}</div>
        <div><b>Saldo:</b> @saldo{0,0}</div>
        <div><b>Segmento:</b> @segmento</div>
    </div>
""", renderers=[renderer])
p_ques4.add_tools(hover)

callback = CustomJS(args=dict(source=source), code="""
    const indices = source.selected.indices;
    if (indices.length > 0) {
        const seg = source.data['segmento'][indices[0]];
        const new_indices = [];
        for (let i = 0; i < source.data['segmento'].length; i++) {
            if (source.data['segmento'][i] === seg) {
                new_indices.push(i);
            }
        }
        source.selected.indices = new_indices;
    }
""")

taptool = p_ques4.select(type=TapTool)
taptool.callback = callback

p_ques4.title.text_color = theme["ACCENT"] 
p_ques4.title.text_font_size = "20pt"        
p_ques4.title.align = "center"  
p_ques4.xaxis.axis_label_text_color = theme["TEXT"]
p_ques4.yaxis.axis_label_text_color = theme["TEXT"]
p_ques4.xaxis.major_label_text_color = theme["TEXT_SOFT"]
p_ques4.yaxis.major_label_text_color = theme["TEXT_SOFT"]
p_ques4.outline_line_color = theme["BORDER"]
p_ques4.xgrid.grid_line_color = theme["GRID"]
p_ques4.ygrid.grid_line_color = theme["GRID"]
p_ques4.xgrid.grid_line_alpha = theme["GRID_ALPHA"]
p_ques4.ygrid.grid_line_alpha = theme["GRID_ALPHA"]
p_ques4.toolbar.logo = None
p_ques4.toolbar.autohide = True

CHARTS["chart_ques4"] = p_ques4
show(p_ques4)
```

```python
import numpy as np
import pandas as pd
from bokeh.plotting import figure
from bokeh.models import ColumnDataSource, Range1d
from bokeh.io import export_png

# ========================================
# GENERAR DATOS ALEATORIOS (Mock Data)
# ========================================
def get_random_temporal_data():
    years = range(2021, 2026)
    quarters = [1, 2, 3, 4]
    x_categories = [f"{y}-Q{q}" for y in years for q in quarters]
    
    regions = ["Región Norte", "Región Sur", "Región Este", "Región Oeste", "Región Central"]
    
    data = {"period": x_categories, "anio": x_categories, "regions": regions}
    
    np.random.seed(42) 
    total_abs_list = []
    
    for x in x_categories:
        tot = np.random.randint(10000, 50000)
        total_abs_list.append(tot)

    data["total"] = total_abs_list

    for r in regions:
        region_vals = []
        region_pcts = []
        for tot in total_abs_list:
            pct = np.random.uniform(0.1, 0.4)
            log_val = np.log(tot + 1) * pct
            region_vals.append(round(log_val, 6))
            region_pcts.append(round(pct * 100, 2))
            
        data[r] = region_vals
        data[f"{r}_pct"] = region_pcts

    return data

# ========================================
# CREAR Y EXPORTAR CHART (PNG LIMPIO)
# ========================================
def generate_temporal_png(temporal_data, output_filename="portada_grafica.png"):
    years = temporal_data.get("anio", [])
    regions = temporal_data.get("regions", [])
    totals = temporal_data.get("total", [])

    x_vals = years if len(years) > 0 else ["-"]

    bars_source = ColumnDataSource(data={"anio": x_vals, **{r: temporal_data.get(r, [0] * len(x_vals)) for r in regions}})
    
# Normalizamos la línea y la elevamos un poco más para que luzca bien sobre las barras
    max_tot = max(totals) if totals and max(totals) > 0 else 1
    normalized_totals = [t / max_tot * 20 for t in totals] if totals else [0] * len(x_vals)
    
    line_source = ColumnDataSource(data={"anio": x_vals, "total_norm": normalized_totals})

    gold_palette = [
        "#1a1a1a", "#222222", "#2c2418",
        "#3a3120", "#4a3b1d", "#5c4a1f", "#6d5422"
    ]
    colors = [gold_palette[i % len(gold_palette)] for i in range(len(regions))]

    p = figure(
        x_range=x_vals,
        height=600,       
        width=1000,       
        toolbar_location=None, 
        background_fill_color=theme["BACKGROUND"],
        border_fill_color=theme["BACKGROUND"],
        min_border=0,             # <-- ELIMINA EL MARGEN INTERNO POR DEFECTO DE BOKEH
        min_border_left=0,
        min_border_right=0,
        min_border_top=0,
        min_border_bottom=0
    )

    p.y_range = Range1d(0, 1)

    # ========================================
    # OCULTAR COMPLETAMENTE TODO LO INNECESARIO
    # ========================================

    p.xaxis.visible = False
    p.yaxis.visible = False
    p.xgrid.grid_line_color = None
    p.ygrid.grid_line_color = None
    p.outline_line_color = None      # <-- Elimina el borde exterior del gráfico
    p.border_fill_color = None       # <-- Quita el color del panel de bordes para que sea transparente
    p.background_fill_color = theme["BACKGROUND"]

    # Barras apiladas
    p.vbar_stack(
        regions, x="anio", width=0.5, color=colors,
        line_color=theme["POSITIVE"], border_radius=100, line_width=0.7,
        source=bars_source
    )

    # Línea de tendencia (usando el eje principal normalizado)
    p.line(
        x='anio', y='total_norm', source=line_source, line_width=5,
        color=theme["ACCENT"], alpha=1
    )
    p.scatter(
        x='anio', y='total_norm', source=line_source, size=15,
        fill_color=theme["TEXT"], line_color=theme["ACCENT"],
        line_width=5
    )

    # Ajuste inicial del rango izquierdo
    try:
        ds = bars_source.data
        if len(x_vals) > 0 and len(regions) > 0:
            sum_logs = np.zeros(len(x_vals))
            for r in regions:
                col = np.array(ds.get(r, [0] * len(x_vals)), dtype=float)
                sum_logs += col
            max_sum = float(np.nanmax(sum_logs)) if sum_logs.size > 0 else 1.0
            p.y_range.start = 0
            p.y_range.end = max_sum * 1.15
    except Exception:
        p.y_range.start = 0
        p.y_range.end = 1
        
    # Exportar el gráfico a PNG (con timeout ampliado para evitar fallos de Selenium)
    export_png(p, filename=output_filename, timeout=15)
    print(f"Gráfica guardada exitosamente en {output_filename}")

# ========================================
# EJECUCIÓN DIRECTA
# ========================================
if __name__ == "__main__":
    fake_data = get_random_temporal_data()
    generate_temporal_png(fake_data, "../charts/portada_grafica.png")
```
