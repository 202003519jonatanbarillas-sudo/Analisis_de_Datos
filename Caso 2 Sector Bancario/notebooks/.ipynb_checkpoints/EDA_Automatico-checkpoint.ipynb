```python
import pandas as pd
import numpy as np
import os
import logging
from datetime import datetime
from bokeh.resources import CDN
from bokeh.plotting import figure
from bokeh.embed import components
from bokeh.models import ColumnDataSource, HoverTool, LinearColorMapper, ColorBar
from bokeh.palettes import RdBu11
from bokeh.layouts import gridplot
import scipy.stats as stats
from scipy.stats import chi2_contingency
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

def calcular_cramers_v(x, y):
    confusion_matrix = pd.crosstab(x, y)
    if confusion_matrix.size == 0: return 0.0
    chi2, _, _, _ = chi2_contingency(confusion_matrix)
    n = confusion_matrix.sum().sum()
    if n <= 1: return 0.0
    phi2 = chi2 / n
    r, k = confusion_matrix.shape
    phi2corr = max(0, phi2 - ((k-1)*(r-1))/(n-1))
    rcorr = r - ((r-1)**2)/(n-1)
    kcorr = k - ((k-1)**2)/(n-1)
    if rcorr <= 1 or kcorr <= 1: return 0.0
    denominator = min(kcorr - 1, rcorr - 1)
    if denominator <= 0: return 0.0
    return np.sqrt(phi2corr / denominator)

def calcular_correlation_ratio(categories, measurements):
    fcat, _ = pd.factorize(categories)
    cat_array = np.array(fcat)
    meas_array = np.array(measurements, dtype=float)
    cat_s = np.unique(cat_array)
    total_mean = np.mean(meas_array)
    n = len(meas_array)
    if n == 0: return 0.0
    ss_between = sum(len(meas_array[cat_array == cat]) * (np.mean(meas_array[cat_array == cat]) - total_mean)**2 for cat in cat_s)
    ss_total = sum((meas_array - total_mean)**2)
    if ss_total == 0: return 0.0
    return np.sqrt(ss_between / ss_total)

def generar_matriz_asociaciones(df):
    todas_cols = list(df.select_dtypes(include=["float64", "int64"]).columns) + \
                 list(df.select_dtypes(include=["object", "str"]).columns)
    
    if len(todas_cols) < 2: return None
        
    matriz = pd.DataFrame(index=todas_cols, columns=todas_cols, dtype=float)
    num_cols = set(df.select_dtypes(include=["float64", "int64"]).columns)
    cat_cols = set(df.select_dtypes(include=["object", "str"]).columns)
    
    for c1 in todas_cols:
        for c2 in todas_cols:
            if c1 == c2:
                matriz.loc[c1, c2] = 1.0
            elif c1 in num_cols and c2 in num_cols:
                matriz.loc[c1, c2] = df[c1].corr(df[c2])
            elif c1 in cat_cols and c2 in cat_cols:
                valid_data = df[[c1, c2]].dropna()
                matriz.loc[c1, c2] = calcular_cramers_v(valid_data[c1], valid_data[c2]) if len(valid_data) > 0 else 0.0
            else:
                valid_data = df[[c1, c2]].dropna()
                if len(valid_data) > 0:
                    if c1 in cat_cols and c2 in num_cols:
                        matriz.loc[c1, c2] = calcular_correlation_ratio(valid_data[c1], valid_data[c2])
                    else:
                        matriz.loc[c1, c2] = calcular_correlation_ratio(valid_data[c2], valid_data[c1])
                else:
                    matriz.loc[c1, c2] = 0.0
    return matriz.fillna(0)

def clasificar_estado(porcentaje, bajo, revisar):
    if porcentaje == 0: return "OK"
    elif porcentaje <= bajo: return "Bajo"
    elif porcentaje <= revisar: return "Revisar"
    else: return "Crítico"

def revisar_nulos(df, bajo=5, revisar=10):
    total_filas = len(df)
    nulos = df.isnull().sum()
    no_nulos = total_filas - nulos 
    porcentaje = (nulos / total_filas) * 100
    
    reporte_nulos = pd.DataFrame({
        "Columna": nulos.index, 
        "No Nulos": no_nulos.values, 
        "Nulos": nulos.values, 
        "Porcentaje": porcentaje.values.round(2)
    })
    
    reporte_nulos["Estado"] = [clasificar_estado(p, bajo, revisar) for p in reporte_nulos["Porcentaje"]]
    
    return reporte_nulos

def revisar_duplicados(df):
    cantidad = df.duplicated().sum()
    porcentaje = (cantidad / len(df)) * 100
    return pd.DataFrame({"Duplicados Totales": [cantidad], "Porcentaje": [round(porcentaje, 2)]})

def revisar_columnas_constantes(df):
    constantes = [col for col in df.columns if df[col].nunique() == 1]
    return pd.DataFrame({"Columnas Constantes": constantes})

def chunk_list(lst, n):
    for i in range(0, len(lst), n):
        yield lst[i:i + n]

def calcular_quality_score(métricas):
    score = 100
    if métricas['porcentaje_nulos'] > 0: score -= min(20, métricas['porcentaje_nulos'] * 2)
    if métricas['porcentaje_duplicados'] > 0: score -= min(15, métricas['porcentaje_duplicados'] * 5)
    if métricas['cols_constantes'] > 0: score -= min(10, métricas['cols_constantes'] * 2)
    if métricas['outliers_totales'] > 0: score -= min(15, (métricas['outliers_totales']/métricas['filas']) * 100)
    
    score = max(0, round(score))
    
    if score >= 90: clasificacion = "Excelente"
    elif score >= 80: clasificacion = "Muy Buena"
    elif score >= 60: clasificacion = "Regular"
    else: clasificacion = "Deficiente"
    
    return score, clasificacion

def generar_recomendaciones(métricas, corr_matrix=None):
    recomendaciones = []
    
    if métricas['porcentaje_nulos'] > 5:
        recomendaciones.append("<strong>Alta presencia de nulos:</strong> Se recomienda evaluar técnicas de imputación avanzadas (KNN, IterativeImputer) en lugar de eliminación para no perder varianza.")
    elif métricas['porcentaje_nulos'] > 0:
        recomendaciones.append("<strong>Valores faltantes leves:</strong> Existen datos nulos. Considere imputación por media/mediana o interpolación según el caso de negocio.")
        
    if métricas['porcentaje_duplicados'] > 1:
        recomendaciones.append("<strong>Registros duplicados:</strong> Se sugiere purgar los registros duplicados para evitar sesgos en el entrenamiento de posibles modelos predictivos.")
        
    if métricas['cols_constantes'] > 0:
        recomendaciones.append("<strong>Variables de varianza cero:</strong> Existen columnas constantes. Estas variables no aportan información predictiva y pueden ser eliminadas.")
        
    if métricas['errores_logicos'] > 0:
        recomendaciones.append("<strong>Errores lógicos detectados:</strong> Se encontraron valores negativos o en cero donde habitualmente no corresponden. Revisar el proceso de ingesta de datos.")
        
    if corr_matrix is not None:
        # Detectar multicolinealidad
        upper_tri = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
        high_corr = [column for column in upper_tri.columns if any(upper_tri[column] > 0.85)]
        if high_corr:
            recomendaciones.append(f"<strong>Multicolinealidad detectada:</strong> Las variables {', '.join(high_corr)} presentan una correlación superior a 0.85. Se recomienda aplicar PCA o eliminar variables redundantes.")
            
    if not recomendaciones:
        recomendaciones.append("<strong>Dataset estructurado:</strong> El dataset presenta condiciones óptimas preliminares. Listo para fases de modelado.")
        
    html = ""
    for rec in recomendaciones:
        html += f"<div class='recommendation-alert'><i class='fas fa-lightbulb'></i> {rec}</div>"
    return html

def generar_reporte_eda(ruta_archivo, nombre_salida="reporte_eda.html", autor="Javier Barillas"):
    logger.info(f"Leyendo archivo: {ruta_archivo}...")
    
    if ruta_archivo.endswith('.csv'):
        df = pd.read_csv(ruta_archivo)
    elif ruta_archivo.endswith(('.xls', '.xlsx')):
        df = pd.read_excel(ruta_archivo)
    else:
        raise ValueError("Formato no soportado. Usa CSV o Excel.")

    logger.info("Procesando métricas base...")
    
    filas, columnas = df.shape
    cols_numericas = df.select_dtypes(include=["float64", "int64"]).columns
    cols_categoricas = df.select_dtypes(include=["object", "str"]).columns
    cols_booleanas = df.select_dtypes(include=["bool"]).columns
    
    nulos_totales = df.isnull().sum().sum()
    porcentaje_nulos = round((nulos_totales / df.size) * 100, 2)
    duplicados = df.duplicated().sum()
    porcentaje_duplicados = round((duplicados / filas) * 100, 2)
    memoria_mb = round(df.memory_usage(deep=True).sum() / (1024**2), 2)
    constantes_df = revisar_columnas_constantes(df)
    num_constantes = len(constantes_df)

    # --- CÁLCULO DE OUTLIERS Y ERRORES ---
    outliers_totales = 0
    errores_logicos = 0
    corr_matrix = None
    
    if len(cols_numericas) > 0:
        Q1 = df[cols_numericas].quantile(0.25)
        Q3 = df[cols_numericas].quantile(0.75)
        IQR = Q3 - Q1
        outliers_count = ((df[cols_numericas] < (Q1 - 1.5 * IQR)) | (df[cols_numericas] > (Q3 + 1.5 * IQR))).sum()
        outliers_totales = outliers_count.sum()
        errores_logicos = (df[cols_numericas] < 0).sum().sum()
        corr_matrix = df[cols_numericas].corr()

    metricas = {
        'filas': filas, 'porcentaje_nulos': porcentaje_nulos, 
        'porcentaje_duplicados': porcentaje_duplicados, 'cols_constantes': num_constantes,
        'outliers_totales': outliers_totales, 'errores_logicos': errores_logicos
    }
    
    score, clasificacion_score = calcular_quality_score(metricas)

    secciones = []

    fecha_actual = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    header_html = f"""
    <div class="dashboard-header">
        <h1 style="margin:0; border:none; padding:0; color: white;"><i class="fas fa-chart-line"></i> Reporte Exploratorio de Datos Avanzado</h1>
        <p style="margin-top:10px; font-size:1.1em; opacity: 0.9;">Generado automáticamente por el pipeline de análisis</p>
        <div style="display: flex; justify-content: center; gap: 20px; margin-top: 15px; font-size: 0.9em; opacity: 0.8;">
            <span><i class="fas fa-file-alt"></i> Archivo: {os.path.basename(ruta_archivo)}</span>
            <span><i class="fas fa-clock"></i> Fecha: {fecha_actual}</span>
            <span><i class="fas fa-user-circle"></i> Autor: {autor}</span>
        </div>
    </div>
    """
    secciones.append(header_html)

    kpi_html = f"""
    <div class="kpi-container">
        <div class="kpi-card"><i class="fas fa-table"></i><h4>Registros</h4><div class="value">{filas:,}</div></div>
        <div class="kpi-card"><i class="fas fa-columns"></i><h4>Variables</h4><div class="value">{columnas:,}</div></div>
        <div class="kpi-card"><i class="fas fa-hashtag"></i><h4>Numéricas</h4><div class="value">{len(cols_numericas)}</div></div>
        <div class="kpi-card"><i class="fas fa-font"></i><h4>Categóricas</h4><div class="value">{len(cols_categoricas)}</div></div>
        <div class="kpi-card"><i class="fas fa-ban"></i><h4>% Nulos</h4><div class="value">{porcentaje_nulos}%</div></div>
        <div class="kpi-card"><i class="fas fa-copy"></i><h4>Duplicados</h4><div class="value">{duplicados:,}</div></div>
        <div class="kpi-card"><i class="fas fa-hdd"></i><h4>Memoria Usada</h4><div class="value">{memoria_mb} MB</div></div>
    </div>
    """
    secciones.append(kpi_html)

    logger.info("Generando resumen de estructura y vista previa...")
    
    df_estructura = pd.DataFrame({
        'Columna': df.columns,
        'Tipo de Dato': df.dtypes.astype(str),
        'Ejemplo (Primer Registro)': [str(df[col].iloc[0]) if len(df) > 0 else "" for col in df.columns]
    })
    
    html_estructura = (
        df_estructura.style
        .set_table_attributes('class="styled-table"')
        .to_html(index=False, escape=False)
    )
    
    df_preview = df.head(5)
    html_preview = (
        df_preview.style
        .set_table_attributes('class="styled-table"')
        .format(precision=2)
        .to_html(index=False, escape=False)
    )
    secciones.append("<div class='section-card'>")
    secciones.append("<div class='section-title'><i class='fas fa-list-alt'></i> Estructura y Tipos de Datos del Dataset</div>")
    secciones.append("<div style='overflow-x: auto; padding: 5px 0;'>")
    secciones.append(html_estructura)
    secciones.append("</div></div>")
    secciones.append("<div class='section-card'>")
    secciones.append("<div class='section-title'><i class='fas fa-table'></i> Vista Previa del Contenido (Primeras 5 Filas)</div>")
    secciones.append("<div style='overflow-x: auto; padding: 5px 0;'>")
    secciones.append(html_preview)
    secciones.append("</div></div>")


    color_score = "#27ae60" if score >= 80 else "#f39c12" if score >= 60 else "#e74c3c"
    color_score = "#27ae60" if score >= 80 else "#f39c12" if score >= 60 else "#e74c3c"
    
    resumen_text = f"El dataset analizado consta de <strong>{filas:,} registros</strong> y <strong>{columnas:,} variables</strong> (compuestas por {len(cols_numericas)} numéricas, {len(cols_categoricas)} categóricas y {len(cols_booleanas)} booleanas). "
    resumen_text += f"Durante la fase de auditoría preliminar, se detectó un <strong>{porcentaje_nulos}% de valores nulos</strong> a nivel global. "
    resumen_text += f"Se identificaron {duplicados:,} filas duplicadas y {num_constantes} variables con valores constantes. "
    if outliers_totales > 0:
        resumen_text += f"A nivel estadístico, se detectaron <strong>{outliers_totales:,} valores atípicos</strong> (outliers) utilizando el método IQR y {errores_logicos} posibles errores lógicos (valores negativos no esperados)."
    
    resumen_html = f"""
    <div style="display: grid; grid-template-columns: 2fr 1fr; gap: 20px; margin-bottom: 30px;">
        <div class="section-card" style="margin-bottom:0;">
            <div class="section-title"><i class="fas fa-file-signature"></i> Executive Summary</div>
            <p style="line-height: 1.6; color: #555;">{resumen_text}</p>
        </div>
        <div class="section-card" style="margin-bottom:0; text-align: center; display: flex; flex-direction: column; justify-content: center;">
            <div class="section-title" style="justify-content: center;"><i class="fas fa-tachometer-alt"></i> Data Quality Score</div>
            <div style="font-size: 3.5em; font-weight: bold; color: {color_score};">{score}/100</div>
            <div style="color: #7f8c8d; font-weight: bold; text-transform: uppercase; margin-top: 5px;">Clasificación: {clasificacion_score}</div>
        </div>
    </div>
    """
    secciones.append(resumen_html)

    logger.info("Analizando calidad de datos...")
    secciones.append("<div class='section-card'>")
    secciones.append("<div class='section-title'><i class='fas fa-search-plus'></i> Auditoría y Calidad de Datos</div>")
    secciones.append("<div style='display: grid; grid-template-columns: 1fr 1fr; gap: 20px;'>")
    secciones.append("<div><h4>Valores Nulos por Columna</h4>")
    secciones.append(revisar_nulos(df).to_html(index=False, classes="styled-table"))
    secciones.append("</div>")
    secciones.append("<div><h4>Filas Duplicadas</h4>")
    secciones.append(revisar_duplicados(df).to_html(index=False, classes="styled-table"))

    if not constantes_df.empty:
        secciones.append("<h4 style='margin-top:20px;'>Columnas Constantes</h4>")
        secciones.append(constantes_df.to_html(index=False, classes="styled-table"))
    secciones.append("</div>")
    secciones.append("</div></div>")

    if len(cols_numericas) > 0:
        logger.info("Calculando estadísticas descriptivas avanzadas...")
        stats_avanzadas = pd.DataFrame(index=cols_numericas)
        stats_avanzadas['Media'] = df[cols_numericas].mean()
        stats_avanzadas['Mediana'] = df[cols_numericas].median()
        stats_avanzadas['Desv. Est.'] = df[cols_numericas].std()
        stats_avanzadas['Mínimo'] = df[cols_numericas].min()
        stats_avanzadas['Máximo'] = df[cols_numericas].max()
        stats_avanzadas['Asimetría (Skew)'] = df[cols_numericas].skew()
        stats_avanzadas['Curtosis'] = df[cols_numericas].kurtosis()
        ceros_count = (df[cols_numericas] == 0).sum()
        negativos_count = (df[cols_numericas] < 0).sum()        
        stats_avanzadas['Ceros (Count)'] = ceros_count
        stats_avanzadas['Negativos (Error)'] = negativos_count
        stats_avanzadas['Outliers (IQR)'] = outliers_count

        format_dict = {
            'Media': '{:,.2f}',
            'Mediana': '{:,.2f}',
            'Desv. Est.': '{:,.2f}',
            'Mínimo': '{:,.2f}',
            'Máximo': '{:,.2f}',
            'Asimetría (Skew)': '{:,.2f}',
            'Curtosis': '{:,.2f}',
            'Ceros (Count)': '{:,.0f}',
            'Negativos (Error)': '{:,.0f}',
            'Outliers (IQR)': '{:,.0f}'
        }

        html_stats = (
            stats_avanzadas.style
            .format(format_dict)
            .apply(lambda x: ['color: #e74c3c; font-weight: bold' if v > 0 else '' for v in x], subset=['Negativos (Error)'])
            .set_table_attributes('class="styled-table"')
            .to_html(escape=False)
        )
        
        secciones.append("<div class='section-card'>")
        secciones.append("<div class='section-title'><i class='fas fa-calculator'></i> Estadística Descriptiva y Anomalías Numéricas</div>")
        secciones.append("<div style='overflow-x: auto; padding: 5px 0;'>")
        secciones.append(html_stats)
        secciones.append("</div></div>")

    if len(cols_numericas) > 1:
        logger.info("Generando mapas interactivos...")
        corr_plot = corr_matrix.iloc[::-1, :]
        corr_plot.index = corr_plot.index.astype(str)
        corr_plot.columns = corr_plot.columns.astype(str)
        corr_plot.index.name = 'AllColumns1'
        corr_plot.columns.name = 'AllColumns2'
        df_corr = corr_plot.stack().rename("value").reset_index()
        df_corr['valor_texto'] = df_corr['value'].apply(lambda x: f"{x:.2f}")
        df_corr['color_texto'] = df_corr['value'].apply(lambda x: "white" if abs(x) > 0.6 else "black")
        source = ColumnDataSource(df_corr)
        mapper = LinearColorMapper(palette=RdBu11, low=-1, high=1)
        
        p_corr = figure(
            title="Heatmap de Pearson", 
            width=750,
            height=600,
            x_range=list(corr_matrix.columns), 
            y_range=list(corr_matrix.columns),
            tools="pan,wheel_zoom,box_zoom,reset,save"
        )
        
        p_corr.rect(
            x="AllColumns2", 
            y="AllColumns1", 
            width=1, 
            height=1, 
            source=source,
            fill_color={'field': 'value', 'transform': mapper}, 
            line_color="white"
        )
        
        p_corr.text(
            x="AllColumns2", 
            y="AllColumns1", 
            text="valor_texto", 
            source=source,
            text_color="color_texto", text_align="center", 
            text_baseline="middle", 
            text_font_size="9pt", 
            text_font_style="bold"
        )
        
        p_corr.grid.grid_line_color = None
        p_corr.axis.axis_line_color = None
        p_corr.axis.major_tick_line_color = None
        p_corr.xaxis.major_label_orientation = np.pi/4
        p_corr.add_layout(ColorBar(color_mapper=mapper, location=(0,0)), 'right')
        p_corr.add_tools(HoverTool(tooltips=[("Variables", "@AllColumns2 vs @AllColumns1"), ("Correlación", "@value{0.00}")]))
        
        script_corr, div_corr = components(p_corr)
        
        secciones.append("<div class='section-card'>")
        secciones.append("<div class='section-title'><i class='fas fa-project-diagram'></i> Matriz de Correlación Interactiva</div>")
        secciones.append(f"<div class='bokeh-container'>{script_corr}\n{div_corr}</div>")
        secciones.append("</div>")

    if len(cols_numericas) > 0:
        logger.info("Construyendo gráficos de distribución...")
        df_plot = df if len(df) <= 50000 else df.sample(50000, random_state=42)
        hist_figures, qq_figures = [], []
        
        for col in cols_numericas:
            data = df_plot[col].dropna()
            if len(data) == 0: continue
            
            # Histograma
            hist, edges = np.histogram(data, bins=30)
            source_hist = ColumnDataSource(data=dict(
                top=hist, bottom=np.zeros_like(hist), left=edges[:-1], right=edges[1:],
                rango=[f"{l:.2f} a {r:.2f}" for l, r in zip(edges[:-1], edges[1:])]
            ))
            p_hist = figure(title=f"Dist: {col}", 
                            width=450, 
                            height=280, 
                            tools="pan,wheel_zoom,reset,save", 
                            background_fill_color="#fafafa"
            )
            
            p_hist.quad(top='top', 
                        bottom='bottom', 
                        left='left', 
                        right='right', 
                        source=source_hist, 
                        fill_color="#3498db", 
                        line_color="white",
                        alpha=0.8
            )
            
            p_hist.add_tools(HoverTool(tooltips=[("Rango", "@rango"), ("Frecuencia", "@top")]))
            hist_figures.append(p_hist)
            osm, osr = stats.probplot(data, dist="norm", fit=False)
            source_qq = ColumnDataSource(data=dict(x=osm, y=osr))
            
            p_qq = figure(title=f"QQ-Plot: {col}", 
                          width=450, 
                          height=280, 
                          tools="pan,wheel_zoom,reset,save", 
                          background_fill_color="#fafafa"
            )
            
            p_qq.scatter('x', 
                         'y', 
                         source=source_qq, 
                         size=4, 
                         color="#2c3e50", 
                         alpha=0.5
            )
            
            min_v, max_v = min(osm.min(), osr.min()), max(osm.max(), osr.max())
            p_qq.line([min_v, max_v], [min_v, max_v], color="#e74c3c", line_dash="dashed", line_width=1.5)
            qq_figures.append(p_qq)

        grid_hist = gridplot(list(chunk_list(hist_figures, 2)), toolbar_location='right')
        grid_qq = gridplot(list(chunk_list(qq_figures, 2)), toolbar_location='right')

        script_h, div_h = components(grid_hist)
        script_q, div_q = components(grid_qq)
        
        secciones.append("<div class='section-card'>")
        secciones.append("<div class='section-title'><i class='fas fa-chart-bar'></i> Distribuciones Univariadas</div>")
        secciones.append(f"<div class='bokeh-container'>{script_h}\n{div_h}</div>")
        secciones.append("</div>")
        
        secciones.append("<div class='section-card'>")
        secciones.append("<div class='section-title'><i class='fas fa-balance-scale'></i> Pruebas de Normalidad (QQ-Plots)</div>")
        secciones.append(f"<div class='bokeh-container'>{script_q}\n{div_q}</div>")
        secciones.append("</div>")

    if len(cols_categoricas) > 0:
        logger.info("Generando gráficos categóricos...")
        secciones.append("<div class='section-card'>")
        secciones.append("<div class='section-title'><i class='fas fa-tags'></i> Frecuencia de Categorías</div>")
        
        cat_figures = []
        mensajes_omitidos = []
        
        for col in cols_categoricas:
            valores_unicos = df[col].nunique()
            if valores_unicos <= 30:
                counts = df[col].value_counts()
                categorias = counts.index.astype(str).tolist()
                frecuencias = counts.values.tolist()
                
                source_cat = ColumnDataSource(data=dict(categorias=categorias, frecuencias=frecuencias))
                p_cat = figure(
                    x_range=categorias, 
                    title=f"Cat: {col}", 
                    width=450, 
                    height=280, 
                    tools="pan,wheel_zoom,box_zoom,reset,save", 
                    background_fill_color="#fafafa"
                )
                
                p_cat.vbar(
                    x='categorias',
                    top='frecuencias', 
                    width=0.8, 
                    source=source_cat, 
                    line_color='white', 
                    fill_color='#e67e22'
                )
                
                p_cat.add_tools(HoverTool(tooltips=[("Categoría", "@categorias"), ("Cantidad", "@frecuencias")]))
                p_cat.xaxis.major_label_orientation = np.pi/4
                cat_figures.append(p_cat)
            else:
                mensajes_omitidos.append(f"<p style='color: #7f8c8d; font-size: 0.9em;'><i class='fas fa-info-circle'></i> Gráfico omitido para '<b>{col}</b>' por alta cardinalidad ({valores_unicos} valores únicos).</p>")

        if len(cat_figures) > 0:
            grid_cat = gridplot(list(chunk_list(cat_figures, 2)), toolbar_location='right')
            script_cat, div_cat = components(grid_cat)
            secciones.append(f"<div class='bokeh-container'>{script_cat}\n{div_cat}</div>")
        
        for msg in mensajes_omitidos:
            secciones.append(msg)
            
        secciones.append("</div>")

    if len(cols_numericas) > 1:
        logger.info("Aplicando Análisis de Componentes Principales...")
        df_pca = df[cols_numericas].dropna()
        if len(df_pca) > 2:
            scaler = StandardScaler()
            scaled_data = scaler.fit_transform(df_pca)
            pca = PCA()
            pca_result = pca.fit_transform(scaled_data)
            var_exp = pca.explained_variance_ratio_ * 100
            
            secciones.append("<div class='section-card'>")
            secciones.append("<div class='section-title'><i class='fas fa-atom'></i> Análisis de Componentes Principales (PCA)</div>")
            secciones.append("<p style='color: #666; margin-bottom: 20px;'>Evaluación de dimensionalidad, varianza explicada y pesos de las variables originales.</p>")
            
            pca_figures = []
            n_comps = len(var_exp)
            comp_labels = [f"PC{i+1}" for i in range(n_comps)]
            source_scree = ColumnDataSource(data=dict(comps=comp_labels, var=var_exp))
            
            p_scree = figure(
                x_range=comp_labels, 
                title="Scree Plot", 
                width=420, 
                height=300, 
                tools="pan,wheel_zoom,reset,save", 
                background_fill_color="#fafafa"
            )
            
            p_scree.vbar(
                x='comps', 
                top='var', 
                width=0.7, 
                source=source_scree, 
                fill_color="#9b59b6", 
                line_color="white"
            )
            
            p_scree.xaxis.major_label_orientation = np.pi/4
            p_scree.yaxis.axis_label = "Varianza Explicada (%)"
            p_scree.add_tools(HoverTool(tooltips=[("Componente", "@comps"), ("Varianza", "@var{0.00}%")]))
            pca_figures.append(p_scree)
            
            # Scatter Plot
            df_scatter = pd.DataFrame(pca_result[:, :2], columns=['PC1', 'PC2'])
            source_scatter = ColumnDataSource(df_scatter)
            
            p_scatter = figure(
                title=f"Scatter: PC1 ({var_exp[0]:.1f}%) vs PC2 ({var_exp[1]:.1f}%)", 
                width=420, 
                height=300, 
                tools="pan,wheel_zoom,reset,save", 
                background_fill_color="#fafafa"
            )
            
            p_scatter.scatter(
                'PC1',
                'PC2', 
                source=source_scatter, 
                size=6, 
                color="#1abc9c", 
                alpha=0.7
            )
            
            p_scatter.xaxis.axis_label = f"PC1"
            p_scatter.yaxis.axis_label = f"PC2"
            p_scatter.add_tools(HoverTool(tooltips=[("PC1", "@PC1{0.00}"), ("PC2", "@PC2{0.00}")]))
            pca_figures.append(p_scatter)
            
            # Heatmap Cargas
            loadings = pd.DataFrame(pca.components_.T, index=cols_numericas, columns=comp_labels)
            loadings_plot = loadings.iloc[:, :min(4, n_comps)]
            loadings_plot.index.name = 'Variable'
            loadings_plot.columns.name = 'Componente'
            df_load = loadings_plot.stack().rename("value").reset_index()
            df_load['valor_texto'] = df_load['value'].apply(lambda x: f"{x:.2f}")
            df_load['color_texto'] = df_load['value'].apply(lambda x: "white" if abs(x) > 0.4 else "black")
            
            source_load = ColumnDataSource(df_load)
            mapper_load = LinearColorMapper(palette=RdBu11, low=-1, high=1)
            
            p_load = figure(
                title="Heatmap de Cargas (Loadings)", 
                x_range=list(loadings_plot.columns), 
                y_range=list(loadings_plot.index[::-1]), 
                tools="pan,wheel_zoom,reset,save", 
                width=420, 
                height=300, 
                background_fill_color="#fafafa"
            )
            
            p_load.rect(
                x="Componente", 
                y="Variable", 
                width=1, 
                height=1, 
                source=source_load, 
                fill_color={'field': 'value', 'transform': mapper_load}, 
                line_color="white"
            )
            
            p_load.text(
                x="Componente", 
                y="Variable", 
                text="valor_texto", 
                source=source_load, 
                text_color="color_texto", 
                text_align="center", 
                text_baseline="middle", 
                text_font_size="8pt", 
                text_font_style="bold"
            )
            
            p_load.grid.grid_line_color = None
            p_load.axis.axis_line_color = None
            p_load.xaxis.major_label_orientation = np.pi/4
            p_load.add_layout(ColorBar(color_mapper=mapper_load, location=(0,0)), 'right')
            p_load.add_tools(HoverTool(tooltips=[("Var/Comp", "@Variable vs @Componente"), ("Peso", "@value{0.00}")]))
            pca_figures.append(p_load)
            
            grid_pca = gridplot(list(chunk_list(pca_figures, 2)), toolbar_location='right')
            script_pca, div_pca = components(grid_pca)
            secciones.append(f"<div class='bokeh-container'>{script_pca}\n{div_pca}</div>")
            secciones.append("</div>")

    logger.info("Calculando asociaciones no lineales mixtas...")
    matriz_asoc = generar_matriz_asociaciones(df)
    if matriz_asoc is not None and not matriz_asoc.empty:
        matriz_plot = matriz_asoc.iloc[::-1, :]
        matriz_plot.index = matriz_plot.index.astype(str)
        matriz_plot.columns = matriz_plot.columns.astype(str)
        matriz_plot.index.name = 'VariableRow'
        matriz_plot.columns.name = 'VariableCol'
        
        df_asoc = matriz_plot.stack().rename("value").reset_index()
        df_asoc['valor_texto'] = df_asoc['value'].apply(lambda x: f"{x:.2f}")
        df_asoc['color_texto'] = df_asoc['value'].apply(lambda x: "white" if abs(x) > 0.6 else "black")
        
        source_asoc = ColumnDataSource(df_asoc)
        mapper_asoc = LinearColorMapper(palette=RdBu11, low=-1, high=1)
        p_asoc = figure(
            title="Asociaciones Mixtas (Pearson, Cramér's V, Correlation Ratio)", 
            x_range=list(matriz_asoc.columns), 
            y_range=list(matriz_asoc.columns),
            tools="pan,wheel_zoom,box_zoom,reset,save", 
            width=750, 
            height=600, 
            background_fill_color="#fafafa"
        )
        
        p_asoc.rect(
            x="VariableCol", 
            y="VariableRow", 
            width=1, 
            height=1, 
            source=source_asoc, 
            fill_color={'field': 'value', 'transform': mapper_asoc}, 
            line_color="white"
        )
        
        p_asoc.text(
            x="VariableCol", 
            y="VariableRow", 
            text="valor_texto", 
            source=source_asoc, 
            text_color="color_texto", 
            text_align="center", 
            text_baseline="middle", 
            text_font_size="9pt", 
            text_font_style="bold"
        )
        
        p_asoc.grid.grid_line_color = None
        p_asoc.axis.axis_line_color = None
        p_asoc.xaxis.major_label_orientation = np.pi/4
        p_asoc.add_layout(ColorBar(color_mapper=mapper_asoc, location=(0,0)), 'right')
        p_asoc.add_tools(HoverTool(tooltips=[("Variables", "@VariableCol vs @VariableRow"), ("Fuerza", "@value{0.00}")]))
        
        script_asoc, div_asoc = components(p_asoc)
        secciones.append("<div class='section-card'>")
        secciones.append("<div class='section-title'><i class='fas fa-network-wired'></i> Matriz de Asociaciones Mixtas</div>")
        secciones.append(f"<div class='bokeh-container'>{script_asoc}\n{div_asoc}</div>")
        secciones.append("</div>")

    secciones.append("<div class='section-card'>")
    secciones.append("<div class='section-title'><i class='fas fa-robot'></i> Sistema Experto: Recomendaciones Automáticas</div>")
    secciones.append(generar_recomendaciones(metricas, corr_matrix))
    secciones.append("</div>")

    logger.info("Generando documento final HTML...")
    recursos_bokeh = CDN.render() 
    
    html = f"""
    <!DOCTYPE html>
    <html lang="es">
    <head>
        <meta charset="utf-8">
        <title>Reporte EDA Profesional</title>
        {recursos_bokeh}
        <!-- LIBRERÍAS EXTERNAS (FONT AWESOME & GOOGLE FONTS) -->
        <link href="https://cdnjs.cloudflare.com/ajax/libs/font-awesome/6.0.0/css/all.min.css" rel="stylesheet">
        <link href="https://fonts.googleapis.com/css2?family=Inter:wght@300;400;600;700&display=swap" rel="stylesheet">
        
        <!-- ESTILOS CORPORATIVOS -->
        <style>
            :root {{
                --primary-color: #2c3e50;
                --secondary-color: #3498db;
                --accent-color: #e74c3c;
                --bg-color: #f4f7f6;
                --card-bg: #ffffff;
                --text-main: #333333;
            }}
            body {{ font-family: 'Inter', sans-serif; background-color: var(--bg-color); color: var(--text-main); margin: 0; padding: 20px 40px; }}
            .container {{ max-width: 1400px; margin: auto; }}
            
            /* Header */
            .dashboard-header {{ background: linear-gradient(135deg, var(--primary-color) 0%, #1a252f 100%); padding: 40px; border-radius: 12px; text-align: center; color: white; margin-bottom: 30px; box-shadow: 0 10px 20px rgba(0,0,0,0.1); }}
            
            /* KPIs */
            .kpi-container {{ display: grid; grid-template-columns: repeat(auto-fit, minmax(180px, 1fr)); gap: 15px; margin-bottom: 30px; }}
            .kpi-card {{ background: var(--card-bg); padding: 20px; border-radius: 10px; border-bottom: 4px solid var(--secondary-color); text-align: center; box-shadow: 0 4px 6px rgba(0,0,0,0.05); transition: transform 0.2s; }}
            .kpi-card:hover {{ transform: translateY(-5px); }}
            .kpi-card i {{ font-size: 2em; color: var(--secondary-color); margin-bottom: 10px; opacity: 0.8; }}
            .kpi-card h4 {{ margin: 0; font-size: 0.85em; color: #7f8c8d; text-transform: uppercase; letter-spacing: 1px; }}
            .kpi-card .value {{ font-size: 1.8em; font-weight: bold; color: var(--primary-color); margin-top: 10px; }}
            
            /* Secciones */
            .section-card {{ background: var(--card-bg); padding: 30px; border-radius: 12px; margin-bottom: 30px; box-shadow: 0 4px 15px rgba(0,0,0,0.03); }}
            .section-title {{ color: var(--primary-color); font-size: 1.4em; font-weight: 600; border-bottom: 2px solid var(--secondary-color); padding-bottom: 12px; margin-bottom: 20px; display: flex; align-items: center; gap: 12px; }}
            
            /* Tablas */
            .styled-table {{ width: 100%; border-collapse: collapse; font-size: 0.9em; margin-bottom: 20px; }}
            .styled-table th, .styled-table td {{ padding: 12px 15px; text-align: left; border-bottom: 1px solid #e0e0e0; }}
            .styled-table th {{ background-color: #f8f9fa; color: var(--primary-color); font-weight: 600; }}
            .styled-table tbody tr:hover {{ background-color: #f5f7fa; }}
            
            /* Recomendaciones */
            .recommendation-alert {{ padding: 16px 20px; border-left: 5px solid var(--secondary-color); background: #ebf5fb; margin-bottom: 12px; border-radius: 4px; display: flex; align-items: center; gap: 15px; font-size: 1.05em; }}
            .recommendation-alert i {{ color: var(--secondary-color); font-size: 1.2em; }}
            
            /* Bokeh Contenedor */
            .bokeh-container {{ display: flex; justify-content: center; align-items: center; overflow-x: auto; padding: 15px; background: #fff; border-radius: 8px; }}
        </style>
    </head>
    <body>
        <div class="container">
            {''.join(secciones)}
            
            <div style="text-align: center; margin-top: 40px; color: #7f8c8d; font-size: 0.9em;">
                <p>Pipeline de Análisis Automatizado - Herramienta de Ciencia de Datos Módulo EDA Profesional</p>
            </div>
        </div>
    </body>
    </html>
    """
    
    with open(nombre_salida, "w", encoding="utf-8") as f:
        f.write(html)
        
    logger.info(f"¡ÉXITO! Reporte profesional generado en: {os.path.abspath(nombre_salida)}")

if __name__ == "__main__":
    generar_reporte_eda("Caso_1_lim.csv", nombre_salida="EDA_Automatico_caso_1.html")
```
